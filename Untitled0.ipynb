```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load red wine dataset and add 'type' column
red_wine = pd.read_csv('winequality-red.csv', delimiter=';')
red_wine['type'] = 'red'

# Load white wine dataset and add 'type' column
white_wine = pd.read_csv('winequality-white.csv', delimiter=';')
white_wine['type'] = 'white'

# Print the first few rows of each dataset
print("Red Wine Dataset:")
print(red_wine.head())

print("\n\nWhite Wine Dataset:")
print(white_wine.head())

# Concatenate datasets along the rows
merged_data = pd.concat([red_wine, white_wine], axis=0)

# Display the shape and first few rows of the combined dataset
print("\n\nCombined Wine Dataset Shape:", merged_data.shape)
print(merged_data.head())

# Save the combined dataset to a CSV file
merged_data.to_csv('merged_wine.csv', index=False)

# Plot the quality distribution
plt.figure(figsize=(10, 6))
sns.histplot(data=merged_data, x='quality', hue='type', multiple='stack')
plt.xlabel('Wine Quality')
plt.ylabel('Count')
plt.title('Quality Distribution of Red and White Wines')
plt.show()
```
